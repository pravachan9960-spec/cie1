```python
import spacy
from nltk import ngrams
nlp = spacy.load("en_core_web_sm")
text = "AI is simulation of human intelligence by machine. It includes expects system NLP speech recognition and machine vision"
doc = nlp(text)
bigrams = list(ngrams(doc, 2))
print("bigrams")
print(bigrams)

trigrams = list(ngrams(doc, 3))
print("trigrams")
print(trigrams)
```
