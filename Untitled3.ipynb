```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/content/iris.csv")
df
```

```python
plt.bar(df["sepal.length"],df["sepal.width"])
plt.title("bar graph")
plt.xlabel("sepal.length")
plt.ylabel("sepal.width")
plt.show()
```

```python
df.hist(bins=10,figsize=(10,8))
plt.show
```
